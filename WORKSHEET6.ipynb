```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

students=['arin','aditya','chirag','gurleen','kunal']
marks=np.array([[85,78,92,88],[79,82,74,90],[90,85,89,92],[66,75,80,78],[70,68,75,85]])
print(marks)

totalmarks=np.sum(marks,axis=1)
avgmarks=np.mean(marks,axis=1)
subjectwise=np.sum(marks,axis=0)
subjectavg=np.mean(marks,axis=0)
topper=np.argmax(totalmarks)
bottom=np.argmin(totalmarks)

print("total marks: ", totalmarks)
print("average marks: ", avgmarks)
print("subjectwise total marks: ", subjectwise)
print("subjectwise average marks: ", subjectavg)
print("topper of class: ", topper)
print("Bottom performer of the class: ", bottom)


plt.bar(students,totalmarks, color='red')
plt.title('total marks of students')
plt.xlabel('students')
plt.ylabel('total marks')
plt.grid(axis='y')
plt.show()


plt.bar(students,avgmarks, color='skyblue')
plt.title('average marks of students')
plt.xlabel('students')
plt.ylabel('average marks')
plt.grid(axis='y')
plt.show()

subject_label=['maths','physics','chemistry', 'english']

plt.bar(subject_label,subjectwise, color='green')
plt.title('subject wise marks of students')
plt.xlabel('students')
plt.ylabel('subject wise marks')
plt.grid(axis='y')
plt.show()

```

```python
import numpy as np
from scipy.optimize import  curve_fit
import matplotlib.pyplot as plt

def velocity(a,b,c,t):
    return a*t**2 +b*t+c

time_data=np.array([0,1,2,3,4,5])
velocity_data=np.array([2,3.1,7.9,18.2,34.3,56.2])

popt,pcoc=curve_fit(velocity,time_data,velocity_data)
print(popt)

y=velocity(time_data,2.66428572,-2.60999997,2.38571426)
plt.title("velocity vs time")
plt.plot(time_data,velocity_data,'o',label='Original data')
plt.plot(time_data,y,'-',label="new data")
plt.ylabel('velocity')
plt.xlabel('time')
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

year=np.array([2000,2005,2010,2015,2020])
population=np.array([50,55,70,80,90])
correlation_coefficient,_=stats.pearsonr(year,population)
print("pearson correlation coefficient: {correlation_coefficient:.4f}")

year_to=2008
estimated_population=np.interp(year_to,year,population)
print(f"estimated population is {year_to}:{estimated_population:.2f}")

plt.scatter(year,population,color='blue',label='actual population')
plt.scatter(year_to,estimated_population,color='red',label='estimated population in {year_to}')

plt.plot(year,population,color='black',linestyle='--',label='population trend')
plt.title('population of town over years')
plt.xlabel("year")
plt.ylabel("population")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

def p(x):
    return 3*x**3 - 5*x**2 + 2*x - 8

coefficients = [3, -5, 2, -8]
roots = np.roots(coefficients)

x = np.linspace(-3, 3, 400)
y = p(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='p(x) = 3x³ - 5x² + 2x - 8')
plt.axhline(0, color='black', lw=0.5, ls='--')
plt.axvline(0, color='black', lw=0.5, ls='--')

for root in roots:
    plt.plot(root, p(root), 'ro')
    plt.annotate(f'({root.real:.2f}, {p(root):.2f})', (root, p(root)), textcoords="offset points", xytext=(0,10), ha='center')

plt.title('Plot of the polynomial p(x)')
plt.xlabel('x')
plt.ylabel('p(x)')
plt.ylim(-10, 10)
plt.grid()
plt.legend()
plt.show()

print("Roots of the polynomial are:", roots)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime

def generate_random_text(size_mb):
    num_chars = size_mb * 1024 * 1024
    return 'random ' * (num_chars // 7)

def convert_to_uppercase(text):
    return text.upper()

def measure_conversion_time(text):
    start_time = datetime.now()
    convert_to_uppercase(text)
    end_time = datetime.now()
    return (end_time - start_time).total_seconds()

sizes = [200, 400, 600, 800, 1000]
times = []
for size in sizes:
    random_text = generate_random_text(size)
    elapsed_time = measure_conversion_time(random_text)
    times.append(elapsed_time)
    print(f'Time taken to convert {size}MB text to uppercase: {elapsed_time:.2f} seconds')

plt.figure(figsize=(10, 6))
plt.plot(sizes, times, marker='o')
plt.title('Time taken to convert random text to uppercase')
plt.xlabel('Text Size (MB)')
plt.ylabel('Time (seconds)')
plt.xticks(sizes)
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

def f(x):
    return x**4 - 3*x**3 + 2

initial_guesses = np.array([-2, 0, 1, 2, 3])
minima = []

for guess in initial_guesses:
    result = minimize(f, guess)
    if result.success:
        minima.append(result.x[0])

minima = list(set(np.round(minima, 6)))

x = np.linspace(-2, 3, 400)
y = f(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='f(x) = x^4 - 3x^3 + 2')
plt.axhline(0, color='black', lw=0.5, ls='--')
plt.axvline(0, color='black', lw=0.5, ls='--')

for m in minima:
    plt.plot(m, f(m), 'ro')
    plt.annotate(f'({m:.2f}, {f(m):.2f})', (m, f(m)), textcoords="offset points", xytext=(0,10), ha='center')

plt.title('Plot of the function f(x) and its local minima')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.ylim(-5, 5)
plt.grid()
plt.legend()
plt.show()

print("Local minima found at:", minima)
```
